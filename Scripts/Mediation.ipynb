```python
import numpy as np
import pandas as pd
import scipy.io as sio
from scipy.stats import rankdata
import pingouin as pg
import math
# =========================
# choose file
# =========================
mat_path = ""
# =========================
# load .mat
# =========================
mat = sio.loadmat(mat_path)

X_raw = np.asarray(mat["X_raw"], dtype=float).squeeze()
M_raw = np.asarray(mat["M_raw"], dtype=float).squeeze()
Y_raw = np.asarray(mat["Y_raw"], dtype=float).squeeze()     

# =========================
# rank transform
# =========================
X = rankdata(X_raw, method="average")
M = rankdata(M_raw, method="average")
Y = rankdata(Y_raw, method="average")

# =========================
# make dataframe
# =========================
df = pd.DataFrame({
    "X": X,
    "M": M,
    "Y": Y
})

# =========================
# mediation analysis
# =========================
stats = pg.mediation_analysis(
    data=df,
    x="X",
    m="M",
    y="Y",
    alpha=0.1,      # 95% CI
    n_boot=10000,    # bootstrap samples
    seed=123
)

print(stats)


df = pd.DataFrame({"X": X, "M": M, "Y": Y})

# =========================
# path a: M ~ X
# =========================
mod_a = pg.linear_regression(df[["X"]], df["M"])
a = mod_a.loc[mod_a["names"] == "X", "coef"].values[0]

# =========================
# paths b and c': Y ~ X + M
# =========================
mod_bcp = pg.linear_regression(df[["X", "M"]], df["Y"])
b = mod_bcp.loc[mod_bcp["names"] == "M", "coef"].values[0]
c_prime = mod_bcp.loc[mod_bcp["names"] == "X", "coef"].values[0]

# =========================
# path c: Y ~ X
# =========================
mod_c = pg.linear_regression(df[["X"]], df["Y"])
c = mod_c.loc[mod_c["names"] == "X", "coef"].values[0]

# =========================
# indirect effect
# =========================
ab = a * b

def trunc2(x):
    return math.trunc(x * 100) / 100
```
